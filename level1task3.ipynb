```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("2) Stock Prices Data Set.csv")
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df.head(10), x="symbol", y="close")

plt.title("Closing Price of Stocks")
plt.xlabel("Stock Symbol")
plt.ylabel("Closing Price")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("bar_plot.png")
plt.show()
```

```python
df["date"] = pd.to_datetime(df["date"])

stock = df[df["symbol"] == "AAPL"]

plt.figure(figsize=(12, 5))

plt.plot(stock["date"], stock["close"], label="AAPL Closing Price")

plt.title("AAPL Stock Price Trend")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("line_chart.png")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("2) Stock Prices Data Set.csv")
df.head()
```

```python
df["date"] = pd.to_datetime(df["date"])

stock = df[df["symbol"] == "AAPL"]

plt.figure(figsize=(12, 5))

plt.plot(stock["date"], stock["close"], label="AAPL Closing Price")

plt.title("AAPL Stock Price Trend")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("line_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="open", y="close")

plt.title("Opening Price vs Closing Price")
plt.xlabel("Opening Price")
plt.ylabel("Closing Price")

plt.tight_layout()

plt.savefig("scatter_plot.png")
plt.show()
```
